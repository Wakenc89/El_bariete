```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar el archivo Excel
df = pd.read_excel("anemia_infantil_datos.xlsx", header=3)

print("Datos cargados correctamente")
print("Filas y columnas:", df.shape)

display(df.head())
```

```python
columnas_categoricas = [
    'Departamento',
    'Área',
    'Grupo_Etario',
    'Sexo',
    'Consume_Suplemento_Hierro',
    'Lactancia_Materna_Exclusiva',
    'Nivel_Educativo_Madre',
    'Deteccion_Tardia_Anemia',
    'Grado_Anemia'
]

for columna in columnas_categoricas:
    print(f"\n--- {columna} ---")
    print(df[columna].value_counts(dropna=False))
```

```python
# ============================================
# REVISIÓN DE INCONSISTENCIAS
# ============================================

print("Valores únicos por variable categórica:\n")

columnas_categoricas = [
    'Departamento',
    'Área',
    'Grupo_Etario',
    'Sexo',
    'Consume_Suplemento_Hierro',
    'Lactancia_Materna_Exclusiva',
    'Nivel_Educativo_Madre',
    'Deteccion_Tardia_Anemia',
    'Grado_Anemia'
]

for columna in columnas_categoricas:
    print(f"\n--- {columna} ---")
    print(df[columna].value_counts(dropna=False))
```

```python
# ============================================
# REVISIÓN DE INCONSISTENCIAS
# ============================================

print("Valores únicos por variable categórica:\n")

columnas_categoricas = [
    'Departamento',
    'Área',
    'Grupo_Etario',
    'Sexo',
    'Consume_Suplemento_Hierro',
    'Lactancia_Materna_Exclusiva',
    'Nivel_Educativo_Madre',
    'Deteccion_Tardia_Anemia',
    'Grado_Anemia'
]

for columna in columnas_categoricas:
    print(f"\n--- {columna} ---")
    print(df[columna].value_counts(dropna=False))
```

```python
columnas_numericas = [
    'Edad_Meses',
    'Peso_kg',
    'Talla_cm',
    'Hemoglobina_g_dL'
]

print("Variables numéricas:")
print(columnas_numericas)
```

```python
def detectar_outliers_iqr(data, columna):
    Q1 = data[columna].quantile(0.25)
    Q3 = data[columna].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = data[
        (data[columna] < limite_inferior) |
        (data[columna] > limite_superior)
    ]

    return len(outliers), limite_inferior, limite_superior


print("VALORES ATÍPICOS DETECTADOS MEDIANTE IQR\n")

for columna in columnas_numericas:
    cantidad, limite_inf, limite_sup = detectar_outliers_iqr(
        df, columna
    )

    print(f"{columna}:")
    print(f"  Cantidad de atípicos: {cantidad}")
    print(f"  Límite inferior: {limite_inf:.2f}")
    print(f"  Límite superior: {limite_sup:.2f}")
    print()
```

```python
# ============================================
# BOXPLOTS PARA IDENTIFICAR VALORES ATÍPICOS
# ============================================

plt.figure(figsize=(10, 6))

df[columnas_numericas].boxplot()

plt.title('Detección visual de valores atípicos')
plt.ylabel('Valor')
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig('grafico_7_boxplots_atipicos.png', dpi=150)
plt.show()
```

```python
# ============================================
# DISTRIBUCIÓN DEL GRADO DE ANEMIA
# ============================================

orden_anemia = [
    'Sin anemia',
    'Anemia leve',
    'Anemia moderada',
    'Anemia severa'
]

conteo_anemia = df['Grado_Anemia'].value_counts().reindex(orden_anemia)

plt.figure(figsize=(8, 5))

ax = conteo_anemia.plot(kind='bar')

plt.title('Distribución según grado de anemia')
plt.xlabel('Grado de anemia')
plt.ylabel('Cantidad de niños')
plt.xticks(rotation=0)

for i, valor in enumerate(conteo_anemia):
    porcentaje = valor / len(df) * 100
    ax.text(i, valor + 50, f'{valor}\n({porcentaje:.2f}%)',
            ha='center')

plt.tight_layout()

plt.savefig('grafico_8_grado_anemia.png', dpi=150)
plt.show()
```

```python
# Crear variable objetivo: 1 = tiene anemia, 0 = sin anemia
df['Con_Anemia'] = (df['Grado_Anemia'] != 'Sin anemia').astype(int)

print("Variable 'Con_Anemia' creada correctamente.")
print(df['Con_Anemia'].value_counts())
```

```python
# ============================================
# ANEMIA SEGÚN CONSUMO DE SUPLEMENTO DE HIERRO
# ============================================

tabla_hierro = pd.crosstab(
    df['Consume_Suplemento_Hierro'],
    df['Con_Anemia']
)

tabla_hierro.columns = ['Sin anemia', 'Con anemia']

print("Cantidad de niños según consumo de hierro y presencia de anemia:")
display(tabla_hierro)

print("\nPorcentajes dentro de cada grupo:")
tabla_hierro_porcentaje = pd.crosstab(
    df['Consume_Suplemento_Hierro'],
    df['Con_Anemia'],
    normalize='index'
) * 100

tabla_hierro_porcentaje.columns = ['Sin anemia (%)', 'Con anemia (%)']

display(tabla_hierro_porcentaje.round(2))
```

```python
# ============================================
# GRÁFICO: HIERRO VS ANEMIA
# ============================================

tabla_grafico = pd.crosstab(
    df['Consume_Suplemento_Hierro'],
    df['Con_Anemia']
)

tabla_grafico.columns = ['Sin anemia', 'Con anemia']

ax = tabla_grafico.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Anemia según consumo de suplemento de hierro')
plt.xlabel('Consumo de suplemento de hierro')
plt.ylabel('Cantidad de niños')
plt.xticks(rotation=0)
plt.legend(title='Estado')

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()

plt.savefig('grafico_9_hierro_vs_anemia.png', dpi=150)
plt.show()
```

```python
# ============================================
# PREVALENCIA DE ANEMIA POR GRUPO ETARIO
# ============================================

prevalencia_edad = df.groupby('Grupo_Etario')['Con_Anemia'].mean() * 100

print("Prevalencia de anemia por grupo etario:")
display(prevalencia_edad.round(2))
```

```python
plt.figure(figsize=(9, 5))

ax = prevalencia_edad.plot(kind='bar')

plt.title('Prevalencia de anemia por grupo etario')
plt.xlabel('Grupo etario')
plt.ylabel('Niños con anemia (%)')
plt.xticks(rotation=45)

for i, valor in enumerate(prevalencia_edad):
    ax.text(i, valor + 1, f'{valor:.2f}%',
            ha='center')

plt.tight_layout()

plt.savefig('grafico_10_prevalencia_grupo_etario.png', dpi=150)
plt.show()
```

```python
# ============================================
# MATRIZ DE CORRELACIÓN
# ============================================

df_correlacion = df[
    ['Edad_Meses', 'Peso_kg', 'Talla_cm',
     'Hemoglobina_g_dL', 'Con_Anemia']
]

correlacion = df_correlacion.corr()

print("Matriz de correlación:")
display(correlacion.round(3))
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlacion,
    annot=True,
    fmt='.3f',
    cmap='coolwarm',
    center=0
)

plt.title('Mapa de correlación entre variables')
plt.tight_layout()

plt.savefig('grafico_11_correlacion.png', dpi=150)
plt.show()
```

```python
# ============================================
# HALLAZGO INTERESANTE
# ============================================

grupo_mayor = prevalencia_edad.idxmax()
valor_mayor = prevalencia_edad.max()

grupo_menor = prevalencia_edad.idxmin()
valor_menor = prevalencia_edad.min()

print("HALLAZGO INTERESANTE")
print("--------------------")

print(
    f"El grupo etario con mayor prevalencia de anemia "
    f"es {grupo_mayor}, con {valor_mayor:.2f}%."
)

print(
    f"El grupo etario con menor prevalencia de anemia "
    f"es {grupo_menor}, con {valor_menor:.2f}%."
)
```
